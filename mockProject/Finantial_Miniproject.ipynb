find the ammount of loan given from mont to date

```python
currtnt_date=pd.Timestamp.today()
curretn_month=currtnt_date.month
curretn_day=currtnt_date.day
app_in_current=dp[(dp['issue_date'].dt.month==curretn_month) & (dp['issue_date'].dt.day<=curretn_day)]
sumt=app_in_current['loan_amount'].sum()
num_app_in_current=app_in_current['id'].nunique()
# print(num_app_in_current)
# print(sumt)
print("Total Applications (MTD):", num_app_in_current)
print("Total Loan Amount Funded (MTD):", sumt)
```

convert this in millions 

```python
sumt_millions=sumt/10000000
he=sumt_millions.round(2)
print(f"The Total Ammount of money dispersed in Loans : ${he} Millions")
```

find the good loan 


```python
loan_status=dp.groupby(
    'loan_status'
).agg({
        'id':'count',
    'dti':'mean',
    'loan_amount':'sum',
    'total_payment':'sum'
})
loan_status.rename(columns={'id':'No of Applications'},inplace=True)
display(loan_status)
```

##Good Loan Analysis


```python
# No of Applications

good_loan_app = loan_status.loc['Fully Paid' , 'No of Applications'] + loan_status.loc['Current' , 'No of Applications']

print(good_loan_app)


# Funded Amount

funded_amount = loan_status.loc['Fully Paid' , 'loan_amount'] + loan_status.loc['Current' , 'loan_amount']

print(funded_amount)

# Received Amount

receiver_amt = loan_status.loc['Fully Paid' , 'total_payment'] + loan_status.loc['Current' , 'total_payment']

print(receiver_amt)

# Profit

print(receiver_amt - funded_amount)
```

##Bad Loan Analysis


```python
# No of Applications

Bad_loan_app = loan_status.loc['Charged Off' , 'No of Applications']

print(Bad_loan_app)


# Funded Amount

funded_amount = loan_status.loc['Charged Off' , 'loan_amount']

print(funded_amount)

# Received Amount

receiver_amt = loan_status.loc['Charged Off' , 'total_payment'] 

print(receiver_amt)

# Profit

print(receiver_amt - funded_amount)
```
