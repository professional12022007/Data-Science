```python
class marai:
  def __init__(self,name,age):
    self.name=name
    self.age=age
    self.strength=100
    self.teacher=['A','B','C']
    self.div=['Alpha','Beta']
    self.sub=['DS','DSA','Web Dev']
    print("Constructor")

obj=marai('rammu',34)
print(obj.strength)
print(obj.teacher)
print(obj.div)
print(obj.sub)
```

```python
class student:
  def _init_(self, name, age, div):
    self.name = name
    self.age = age
    self.div = div
a=student('John',20,'gera')
print(a.name)
print(a.age)
print(a.div)
```
