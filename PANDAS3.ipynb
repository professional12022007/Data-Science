```python
import pandas as pd
```

```python
df1=pd.DataFrame({
    'id':[1,2,3,4],
    'name':['Raja','Maini','Rahul','Ghee'],
    'age':[23,24,25,26]
})
df2=pd.DataFrame({
    'id':[3,4,5,6],
    'name':['Rahul','Rani','Rani','Ganda'],
    'age':[42,45,456,56]
})
```

```python
result=pd.merge(df1,df2,on='id',how='outer')
result
# sare ellements a jate hain in this it is like the union of the 2 sets
```

```python
result=pd.merge(df1,df2,on='id',how='inner')
result
# keval common value of both the tables are mentioned based on the both the tables
# like same id in the table
```

```python
result=pd.merge(df1,df2,on='id',how='left')
result
# first the table on the first table is printed and if there is not thesame id on the right table then if it does not exist in the right table then the nul value is assigned to the value
# like same id in the table
```

```python
result=pd.merge(df1,df2,on='id',how='right')
result
# first the table on the second table is printed and if there is not thesame id on the right table then if it does not exist in the right table then the nul value is assigned to the value
# like same id in the table
```

```python
result.rename(columns={'name_x':'tablename_x'},inplace =True)
result
```

```python
result=pd.merge(df1,df2,how='cross')
result
```
