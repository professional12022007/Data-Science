### Using `df.dropna()` to remove missing values

`dropna()` is used to remove rows or columns that contain `NaN` (Not a Number) values. Key parameters include:

*   `axis`: Specifies whether to drop rows (`0` or `'index'`) or columns (`1` or `'columns'`). Default is `0`.
*   `how`: Determines how to consider `NaN` values:
    *   `'any'` (default): Drop if *any* `NaN` values are present in the row/column.
    *   `'all'`: Drop only if *all* values in the row/column are `NaN`.
*   `inplace`: If `True`, modifies the DataFrame in place and returns `None`. If `False` (default), returns a new DataFrame.

Let's create a sample DataFrame with some missing values to demonstrate:

```python
import numpy as np

# Create a sample DataFrame with some NaN values
data_with_nan = {
    'A': [1, 2, np.nan, 4, 5],
    'B': [np.nan, 7, 8, 9, 10],
    'C': [11, 12, 13, np.nan, 15],
    'D': [16, 17, 18, 19, np.nan]
}
df_nan = pd.DataFrame(data_with_nan)
print("Original DataFrame with NaN values:")
display(df_nan)
```

#### 1. Dropping rows with any `NaN` values (default behavior)

```python
# By default, dropna() drops rows that contain *any* NaN values (axis=0, how='any')
df_rows_dropped = df_nan.dropna()
print("DataFrame after dropping rows with any NaN:")
display(df_rows_dropped)
```

#### 2. Dropping columns with any `NaN` values

```python
# To drop columns with any NaN values, set axis=1
df_cols_dropped = df_nan.dropna(axis=1)
print("DataFrame after dropping columns with any NaN:")
display(df_cols_dropped)
```

#### 3. Dropping rows only if all values are `NaN`

```python
# Let's add a row with all NaNs to our sample DataFrame
df_nan.loc[len(df_nan)] = [np.nan, np.nan, np.nan, np.nan]
print("DataFrame with an all-NaN row:")
display(df_nan)

# Now, drop rows only if ALL values are NaN
df_all_nan_rows_dropped = df_nan.dropna(how='all')
print("DataFrame after dropping rows where all values are NaN:")
display(df_all_nan_rows_dropped)
```

### Using `df.sort_values()` to sort a DataFrame

The `sort_values()` method sorts a DataFrame by the values of one or more columns. Key parameters are:

*   `by`: The column name(s) to sort by. Can be a single column name or a list of column names.
*   `axis`: Defaults to 0 (sort by row index labels). For sorting by column values, this is implicitly handled by `by`.
*   `ascending`: Boolean, `True` by default, for ascending order. Can be a single boolean or a list of booleans (one for each column in `by`).
*   `inplace`: If `True`, the DataFrame is modified in place and `None` is returned. If `False` (default), it returns a new sorted DataFrame.

Let's use your main `df` DataFrame for these examples.

#### 1. Sorting by a single column in ascending order (default)

```python
# Sort the DataFrame by 'exam_score' in ascending order
df_sorted_asc = df.sort_values(by='exam_score')
print("DataFrame sorted by 'exam_score' (ascending):")
display(df_sorted_asc.head())
```

#### 2. Sorting by a single column in descending order

```python
# Sort the DataFrame by 'exam_score' in descending order
df_sorted_desc = df.sort_values(by='exam_score', ascending=False)
print("DataFrame sorted by 'exam_score' (descending):")
display(df_sorted_desc.head())
```

#### 3. Sorting by multiple columns

```python
# Sort by 'placement_status' (ascending) then by 'exam_score' (descending)
df_sorted_multi = df.sort_values(by=['placement_status', 'exam_score'], ascending=[True, False])
print("DataFrame sorted by 'placement_status' (asc) then 'exam_score' (desc):")
display(df_sorted_multi.head())
```

#### 4. Using `inplace=True`

```python
# Sort the DataFrame by 'study_hours' in-place
print("Original first 5 rows of df (before inplace sort):")
display(df.head())

df.sort_values(by='study_hours', inplace=True)

print("\nFirst 5 rows of df after inplace sort by 'study_hours':")
display(df.head())
```

### Using `df.sort_index()` to sort a DataFrame by its index

The `sort_index()` method sorts a DataFrame by its index labels rather than by column values. This can be useful for organizing your data based on row or column labels.

Key parameters are:

*   `axis`: Specifies whether to sort by row index (`0` or `'index'`, default) or column index (`1` or `'columns'`).
*   `ascending`: Boolean, `True` by default, for ascending order. Can be a single boolean for the chosen axis.
*   `inplace`: If `True`, the DataFrame is modified in place and `None` is returned. If `False` (default), it returns a new sorted DataFrame.

#### 1. Sorting by row index in ascending order (default)

```python
# Reset the index to default integer index if it was previously sorted by a column, to better show index sorting.
# For demonstration, let's reload the original df to ensure index is not custom sorted.
df_original_index = pd.read_csv('student_data.csv')
df_original_index.rename(columns={"assignments_completed":'assign completed'} , inplace =True)

# Sort the DataFrame by its row index in ascending order (default)
df_index_sorted_asc = df_original_index.sort_index()
print("DataFrame sorted by index (ascending) - head:")
display(df_index_sorted_asc.head())
print("DataFrame sorted by index (ascending) - tail:")
display(df_index_sorted_asc.tail())
```

#### 2. Sorting by row index in descending order

```python
# Sort the DataFrame by its row index in descending order
df_index_sorted_desc = df_original_index.sort_index(ascending=False)
print("DataFrame sorted by index (descending) - head:")
display(df_index_sorted_desc.head())
print("DataFrame sorted by index (descending) - tail:")
display(df_index_sorted_desc.tail())
```

#### 3. Sorting by column index (i.e., column names)

```python
# Sort the DataFrame by its column index (column names) in ascending order
df_cols_sorted_asc = df_original_index.sort_index(axis=1)
print("DataFrame sorted by column names (ascending) - head:")
display(df_cols_sorted_asc.head())
```
